```python
import numpy as np

print("INICIANDO GERAÇÃO DA BASE DE LEITOS (CNES / IBGE)")

# Fixando a semente randômica para garantir reprodutibilidade nos dados gerados
np.random.seed(42)

setores_possiveis = ["UTI Adulto",
                     "UTI Pediátrica",
                     "Enfermaria",
                     "Pronto Socorro",
                     "Centro Obstétrico/Maternidade"]
classificacoes_possiveis = ["Vermelho", "Amarelo", "Verde"]

lista_leitos_reais = []
id_l = 1

# Garantindo a varredura exclusiva sobre a malha de municípios já limpa e tratada
for _, row in df_resumo_nomes.iterrows():
    mun = row['Nome_Municipio']
    cod_gestor = str(row['SP_GESTOR']).strip() # Padroniza como string sem espaços

    # Dimensionamento realista do número de leitos com base nas internações
    qtd_leitos_cidade = max(2, int(row['Total_Internacoes'] / 150000))
    if qtd_leitos_cidade > 15:
        qtd_leitos_cidade = 15

    for _ in range(qtd_leitos_cidade):
        crit = np.random.choice(classificacoes_possiveis, p=[0.2, 0.4, 0.4])
        lista_leitos_reais.append({
            "Leito": f"L-{id_l:03d}",
            "Setores": np.random.choice(setores_possiveis),
            "Classificação": crit,
            "SP_GESTOR": cod_gestor,
            "Nome_Municipio": mun
        })
        id_l += 1

# Conversão para DataFrame do Pandas
df_leitos = pd.DataFrame(lista_leitos_reais)

# Criação do índice numérico de complexidade para ordenação posterior
mapa_complexidade = {'Vermelho': 3, 'Amarelo': 2, 'Verde': 1}
df_leitos['C_Leito'] = df_leitos['Classificação'].map(mapa_complexidade)

# Assegurando o tipo string limpo na coluna de chave estrangeira
df_leitos['SP_GESTOR'] = df_leitos['SP_GESTOR'].astype(str).str.strip()

print(f"✅ Total de leitos reais gerados com sucesso: {len(df_leitos)}")
df_leitos.head()
```

```python
print("INICIANDO GERAÇÃO DA MATRIZ DE FADIGA E EQUIPE")

cargos_possiveis = ['Médico Clínico Geral', 'Médico Intensivista',
                    'Enfermeira Coordenadora', 'Técnica de Enfermagem',
                    'Enfermeiro']
nomes_pool = ['Ana Silva', 'Carlos Souza', 'Mariana Lima', 'Fernando Costa',
              'Beatriz Alves', 'Lucas Mendes', 'Juliana Rocha',
              'Gabriel Martins', 'Camila Santos', 'Rafael Ribeiro']

lista_prof_reais = []
id_p = 1

for _, row in df_resumo_nomes.iterrows():
    mun = row['Nome_Municipio']
    cod_gestor = str(row['SP_GESTOR']).strip()

    # Dimensionamento proporcional de profissionais por município
    qtd_prof_cidade = max(2, int(row['Total_Internacoes'] / 200000))
    if qtd_prof_cidade > 10:
        qtd_prof_cidade = 10

    for _ in range(qtd_prof_cidade):
        horas = np.random.choice([12, 16, 24, 30, 36, 40, 44])
        noites = np.random.choice([0, 1, 2, 3])
        folgas = np.random.choice([0, 1, 2])

        lista_prof_reais.append({
            "ID": id_p,
            "Nome": f"{np.random.choice(nomes_pool)} ({mun[:3]})",
            "Profissão": np.random.choice(cargos_possiveis),
            "Horas semanais": horas,
            "Plantões noturnos": noites,
            "Folgas (Últimos 7 dias)": folgas,
            "SP_GESTOR": cod_gestor,
            "Nome_Municipio": mun
        })
        id_p += 1

df_equipe = pd.DataFrame(lista_prof_reais)

# Função corporativa de cálculo do índice de prontidão e fadiga
def calculo_prontidao(row):
    horas = row['Horas semanais']
    noites = row['Plantões noturnos']
    folgas = row['Folgas (Últimos 7 dias)']

    calculo = 100 - (horas * 1.5) - (noites * 10) + (folgas * 5)
    calculo = max(0, min(100, calculo))

    if calculo >= 70:
        status = '🟢 Apto'
    elif calculo >= 50:
        status = '🟡 Atenção'
    else:
        status = '🔴 Crítico'

    return pd.Series([round(calculo, 1), status])

df_equipe[['Prontidao (%)', 'Status']] = df_equipe.apply(calculo_prontidao, axis=1)
df_equipe['SP_GESTOR'] = df_equipe['SP_GESTOR'].astype(str).str.strip()

print(f"✅ Total de profissionais gerados com sucesso: {len(df_equipe)}")
df_equipe.head()
```

```python
print("EXECUTANDO ALGORITMO DE PAREAMENTO DE LEITOS E ESCALAS")

# Ordenando os leitos por nível de complexidade (do mais crítico ao mais estável)
df_leitos_ordenados = df_leitos.sort_values(by='C_Leito',
                                            ascending=False).reset_index(drop=True)

# Bloqueando profissionais com status crítico
df_equipe_ordenada = df_equipe[df_equipe['Status'] != '🔴 Crítico'].sort_values(
    by='Prontidao (%)', ascending=False).reset_index(drop=True)

pareamento = []
total_leitos = len(df_leitos_ordenados)
total_aptos = len(df_equipe_ordenada)

for i in range(total_leitos):
    if i < total_aptos:
        pareamento.append({
            'ID Leito': df_leitos_ordenados.loc[i, 'Leito'],
            'Setor': df_leitos_ordenados.loc[i, 'Setores'],
            'Classificação': df_leitos_ordenados.loc[i, 'Classificação'],
            'C_Leito': df_leitos_ordenados.loc[i, 'C_Leito'],
            'Profissional Alocado': df_equipe_ordenada.loc[i, 'Nome'],
            'Prontidao (%)': df_equipe_ordenada.loc[i, 'Prontidao (%)'],
            'Status': df_equipe_ordenada.loc[i, 'Status'],
            'Municipio': df_leitos_ordenados.loc[i, 'Nome_Municipio']
        })
    else:
        # Tratamento operacional para alta demanda (dupla cobertura)
        prof_top = df_equipe_ordenada.loc[0, 'Nome']
        pront_top = df_equipe_ordenada.loc[0, 'Prontidao (%)']
        status_top = df_equipe_ordenada.loc[0, 'Status']

        pareamento.append({
            'ID Leito': df_leitos_ordenados.loc[i, 'Leito'],
            'Setor': df_leitos_ordenados.loc[i, 'Setores'],
            'Classificação': df_leitos_ordenados.loc[i, 'Classificação'],
            'C_Leito': df_leitos_ordenados.loc[i, 'C_Leito'],
            'Profissional Alocado': f"{prof_top} (⚠️ Dupla Cobertura)",
            'Prontidao (%)': pront_top,
            'Status': status_top,
            'Municipio': df_leitos_ordenados.loc[i, 'Nome_Municipio']
        })

df_pareamento = pd.DataFrame(pareamento)
print(f"✅ Total de pareamentos executados: {len(df_pareamento)}")
df_pareamento.head(10)
```
